[Cylinder] Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao (kiểu số thực) với hai phương thức:tính diện tích bề mặt (diện tích toàn phần) và thể tích của hình trụ
 
Đầu vào  
Đâu vào từ bàn phím gồm một dòng duy nhất chứa bán kính và chiều cao của hình trụ, cách nhau bởi một dấu cách. 
 
Đầu ra 
In ra màn hình diện tích bề mặt và thể tích của hình trụ (sử dụng lại những hàm vừa viết, π=3.14).  
 
(Kết quả làm tròn đến 2 chữ số sau dấu phẩy). 

```python
class Cylinder:
    def __init__(self, r, h):
        self.bk = r 
        self.cc = h
    def dientich(self):
        return 2*3.14*self.bk**2 + 2*3.14*self.bk*self.cc
    def thetich(self):
        return 3.14*self.bk**2*self.cc
    def __str__(self):
        return f"({self.bk},{self.cc})"
b,c = map(float,input().split())
cylinder1 = Cylinder(b,c)
print(f"Dien tich be mat: {cylinder1.dientich()}:.2f")
print(f"the tich: {cylinder1.thetich()}:.2f")
```

Đầu vào:  Nhập vào tọa độ một điểm M bất kỳ 
 
Đầu ra: 
Thông báo M có phải gốc tọa độ không hoặc có nằm trên trục tung hoặc trục hoành không, khoảng cách từ M tới gốc tọa độ là bao nhiêu (lấy 2 chữ số phần thập phân) 

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def goctd(self):
        return self.x == self.y == 0
    def trucy(self):
        return self.x == 0
    def trucx(self):
        return self.y == 0
    def kc(self, other):
        return ((self.x - other.x)**2 + (self.y - other.y)**2)**0.5
x,y = map(int,input().split())
ham1 = Point(x,y)
ham2 = Point(0,0)
if ham1.goctd():
    print("M nam tren goc toa do")
elif ham1.trucy():
    print("M nam tren truc y")
elif ham1.trucx():
    print("M nam tren truc x")
print(f"{ham1.kc(ham2):.2f}")
```

Xây dựng lớp Rectangle(w, h) với các phương thức: 
 
area() trả về diện tích. 
 
perimeter() trả về chu vi. 
 
scale(k) phóng to hình chữ nhật theo hệ số k (nhân cả w và h), không trả về gì. 
 
Input 
w h k (số thực dương) 
 
Output 
area perimeter sau khi phóng to (làm tròn 2 chữ số thập phân).

```python
class Rectangle:
    def __init__(self ,w ,h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
    def perimeter(self):
        return (self.w + self.h) * 2
    def scale(self ,k):
        self.w = self.w * k
        self.h = self.h * k
w,h,k = map(float,input().split())
n = Rectangle(w,h)
n.scale(k)
print(f"{n.area():.2f}")
print(f"{n.perimeter():.2f}")
```

Yêu cầu 
Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠ 0) với: 
 
Rút gọn phân số trong __init__ (quy ước mẫu > 0). 
 
Các phương thức add/sub/mul/div trả về phân số đã rút gọn. 
 
__str__() in dạng a/b. 
 
Input 
a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*,/}) 
 
Output 
Phân số kết quả đã rút gọn

```python
import math
class Fraction:
    def __init__(self ,num ,den):
        self.tu = num
        self.mau = den
        if self.mau < 0:
            self.tu = -self.tu
            self.mau = -self.mau
        a = math.gcd(self.tu,self.mau)
        self.tu = self.tu // a
        self.mau = self.mau // a
    def __str__(self):
        return f"{self.tu}/{self. mau}"
    def add(self, other):
        return Fraction(self.tu * other.mau + other.tu * self.mau, self.mau * other.mau)
    def sub(self,other):
        return Fraction(self.tu * other.mau - other.tu * self.mau, self.mau * other.mau)
    def mul(self,other):
        return Fraction(self.tu * other.tu,self.mau * other.mau)
    def div(self,other):
        return Fraction(self.tu * other.mau,self.mau * other.tu)
a,b,op,c,d = input().split()
n1 = Fraction(int(a),int(b))
n2 = Fraction(int(c),int(d))
if op == '+':
    print(n1.add(n2))
elif op == '-':
    print(n1.sub(n2))
elif op == '*':
    print(n1.mul(n2))
elif op == '/':
    print(n1.div(n2))
    
```

Lớp BankAccount(owner, balance) với: 
 
deposit(x), withdraw(x); nếu rút quá số dư thì bỏ qua giao dịch. 
 
__str__() in owner: balance. 
 
Input 
Dòng 1: owner initial_balance 
Dòng 2: n (số lệnh) 
Tiếp theo n dòng, mỗi dòng: DEPOSIT x hoặc WITHDRAW x. 
 
Output 
Số dư cuối cùng (in số thực, 2 chữ số). 

```python
class BankAccount:
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = balance
    def tinh(self,other):
        if other.owner == 'Deposit':
            self.balance = self.balance + other.balance
        elif other.owner == 'Withdraw' and self.balance >= other.balance:
            self.balance = self.balance - other.balance
        return self.balance
    def __str__(self):
        return f"{self.balance:.2f}"
x,y = input().split()
n = int(input())
Bank = BankAccount(x,int(y))
for i in range(n):
    a,b = input().split()
    giaodich = BankAccount(a,int(b))
    sodu = Bank.tinh(giaodich)
    Bank = BankAccount(x,int(sodu))
print(f"{sodu:.2f}")
```

Yêu cầu 
Lớp Student(name) với: 
 
add_score(subject, score) 
 
avg() trả về điểm trung bình. 
 
rank() trả về "Excellent" nếu avg ≥ 8, "Good" nếu 6.5–<8, "Average" nếu 5–<6.5, ngược lại "Poor". 
 
Input 
Dòng 1: name 
Dòng 2: n 
n dòng tiếp: subject score (score là số thực 0–10) 
 
Output 
name avg rank (avg 2 chữ số). 

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.score = []
        
    def add_score(self, subject, score):
        return self.score.append(score)
        
    def avg(self):
        a = len(self.score)
        t = 0
        for i in self.score:
            t = t + float(i)
        b = t / a
        return f"{b:.2f}"
        
    def rank(self):
        c = float(self.avg())
        if c >= 8.0:
            return "Excellent"
        elif c >= 6.5:
            return "Good"
        elif c >= 5.0:
            return "Average"
        else:
            return "Poor"

name = input()
n = int(input())

student = Student(name)

for _ in range(n):
    subject, score = input().split()
    student.add_score(subject, float(score))
    
print(name, student.avg(), student.rank())
```
